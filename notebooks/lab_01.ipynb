# Git and GitHub

Git tracks changes locally. GitHub hosts Git repositories online. In this section you will create a branch, inspect a project change, and commit it locally.

### Git cheat sheet

| Command | Purpose |
| --- | --- |
| `git status` | Show the current branch and changes. |
| `git branch` | List local branches. |
| `git switch -c NAME` | Create and switch to a branch. |
| `git add FILE` | Stage a file. |
| `git commit -m "MESSAGE"` | Create a local commit. |
| `git log --oneline` | Show commit history. |
| `git pull` | Get changes from GitHub. |
| `git push` | Send commits to GitHub. |
